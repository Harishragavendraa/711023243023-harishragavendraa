```
import sqlite3
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()


cursor.execute('''
CREATE TABLE sales (
    region VARCHAR(50),
    product VARCHAR(50),
    year INT,
    sales_amount INT
);
''')


cursor.execute('''
INSERT INTO sales VALUES
('North', 'Laptop', 2023, 50000),
('North', 'Mobile', 2023, 30000),
('South', 'Laptop', 2023, 40000),
('South', 'Mobile', 2023, 20000),
('North', 'Laptop', 2024, 60000),
('South', 'Mobile', 2024, 25000);
''')

conn.commit()
print("Database and table 'sales' created, and data inserted successfully.")
```

```
cursor.execute('''
SELECT *
FROM sales
WHERE year = 2023;
''')

results = cursor.fetchall()

for row in results:
    print(row)
```

```
cursor.execute('''
SELECT *
FROM sales
WHERE region = 'North'
  AND product = 'Laptop';
''')

results = cursor.fetchall()

for row in results:
    print(row)
```

```
cursor.execute('''
SELECT region, SUM(sales_amount) AS total_sales
FROM sales
GROUP BY region;
''')

results = cursor.fetchall()

for row in results:
    print(row)
```

```
cursor.execute('''
SELECT region, product, SUM(sales_amount) AS total_sales
FROM sales
GROUP BY region, product

UNION ALL

SELECT region, NULL, SUM(sales_amount)
FROM sales
GROUP BY region

UNION ALL

SELECT NULL, NULL, SUM(sales_amount)
FROM sales;
''')

results = cursor.fetchall()

for row in results:
    print(row)
```

```
cursor.execute('''
SELECT region, product, year, SUM(sales_amount)
FROM sales
GROUP BY region, product, year;
''')

results = cursor.fetchall()

for row in results:
    print(row)
```

```
import pandas as pd

data = {
    'Region': ['North','North','South','South','North','South'],
    'Product': ['Laptop','Mobile','Laptop','Mobile','Laptop','Mobile'],
    'Year': [2023,2023,2023,2023,2024,2024],
    'Sales': [50000,30000,40000,20000,60000,25000]
}

df = pd.DataFrame(data)

# Slice
print(df[df['Year'] == 2023])

# Dice
print(df[(df['Region']=='North') & (df['Product']=='Laptop')])

# Roll-up
print(df.groupby('Region')['Sales'].sum())

# Drill-down
print(df.groupby(['Region','Product','Year'])['Sales'].sum())

# Pivot
print(df.pivot_table(values='Sales', index='Region', columns='Product', aggfunc='sum'))
```
